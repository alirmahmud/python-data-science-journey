```python
company = " genz-ryan   "
active_start = 1500
new_signups = 120
cancellations = 46
price = 189.0
```

```python
## This is for #1
ending_members = active_start + new_signups - cancellations
churn_rate = cancellations/active_start * 100
net_growth = (ending_members - active_start) / active_start * 100
annual_revenue = price * ending_members

print(f"=== {company.strip().upper()} MONTHLY REPORT ===")
print(f"Starting members: {active_start:,}")
print(f"New sign-ups: {new_signups}")
print(f"Cancellations: {cancellations}")
print(f"Ending members: {ending_members:,}")
print(f"Churn rate: {churn_rate:,.2f}%")
print(f"Net growth: {net_growth:,.2f}%")
print(f"Annual revenue: ${annual_revenue:,.2f}")
```
