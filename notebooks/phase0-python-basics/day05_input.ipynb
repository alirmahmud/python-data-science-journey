```python
#This is for exercise 5.2
membership_price = float(input("Input price per member "))
num_members = int(input("Input number of members "))
revenue = membership_price * num_members
print(f"Membership Price: $ {membership_price}") #Membership Price
print(f"Number of members: {num_members}") #Number of Members
print(f"The total revenue is ${revenue:,.2f}")
```

```python
#This is for exercise 5.3
cancels = int(input("Input number of cancellations"))
churn_rate = cancels / num_members * 100
print(f"The number of members cancelled is {cancellations}")
print(f"The churn rate is {churn_rate:,.2f}%")
```
