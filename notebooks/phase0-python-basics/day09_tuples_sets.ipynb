```python
calls = [101, 102, 101, 103, 102]    # account IDs that called today
unique = set(calls)
print(unique)                        # {101, 102, 103}
print(len(unique))                   # 3 different accounts called
print(sorted(unique))                # [101, 102, 103]  (sorted() gives back a list)
```

```python
unique.add(104)          # add one item
unique.add(101)          # already there: nothing happens
unique.discard(999)      # remove if present; no error if it's missing
print(unique)            # the order may look random, and that's normal
print(102 in unique)     # True  (checking a set is very fast)
```

```python
jan = {101, 102, 103}     # members active in January
feb = {102, 103, 104}     # members active in February

print(jan & feb)          # {102, 103}            in BOTH (stayed)
print(jan | feb)          # {101, 102, 103, 104}  in EITHER (everyone)
print(jan - feb)          # {101}                 in Jan, not Feb (churned)
print(feb - jan)          # {104}                 in Feb, not Jan (new)
```

```python
#Exercise 9.4
q1 = {201, 202, 203, 204, 205}
q2 = {203, 204, 205, 206, 207}

#members who stayed in both
print(q1 & q2)

#members who churned (in q1, not q2)
print(q1 - q2)

#new members (in q2, not q1)
print(q2 - q1)
```

```python
#Exercise 9.5
#9.5 Using 9.4, calculate the churn rate: churned ÷ q1 members × 100. Print it like Churn: 40.00%. Hint: len() works on sets.

lenq1 = len(q1)
lenq2 = len(q2)
churned = len(q1 - q2)

churn_rate = churned/lenq1 * 100
print(churn_rate)
print(f"Churn: {churn_rate:,.2f}%.")
```
